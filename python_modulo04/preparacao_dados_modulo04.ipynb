## Leitura e primeira organização da base

Daqui em diante a base bruta é lida, tem as colunas padronizadas e passa por uma primeira conferência, preparando o terreno para o diagnóstico de qualidade e as transformações das próximas seções.

```python
# Passo 4 — Ler o CSV da PRF testando mais de um encoding
#
# utf-8 é tentado antes de latin1 de propósito: latin1 aceita praticamente
# qualquer sequência de bytes sem travar, então se ele viesse primeiro os
# acentos sairiam corrompidos sem nenhum erro aparecer no processo.
# A lógica de fallback continua a mesma do original.

def ler_csv_prf(caminho, sep=";", encodings=("utf-8", "utf-8-sig", "latin1")):
    erro_capturado = None
    for codificacao in encodings:
        try:
            print(f"Tentando leitura com encoding={codificacao}...")
            return pd.read_csv(caminho, sep=sep, encoding=codificacao, low_memory=False)
        except Exception as erro:
            erro_capturado = erro
            print(f"Falhou com {codificacao}: {erro}")
    raise erro_capturado

df = ler_csv_prf(ARQUIVO_BRUTO, sep=SEPARADOR)
print(df.head())

print("Leitura inicial concluída com sucesso.")
```

```python
# Passo 5 — Padronizar os nomes das colunas
# Deixa tudo em minúsculo, sem acento e sem espaço, pra facilitar referenciar
# as colunas no resto do notebook.

def normalizar_nome_coluna(nome):
    nome = str(nome).strip().lower()
    nome = unicodedata.normalize("NFKD", nome).encode("ascii", "ignore").decode("utf-8")
    nome = nome.replace(" ", "_").replace("-", "_").replace("/", "_")
    while "__" in nome:
        nome = nome.replace("__", "_")
    return nome.strip("_")

df.columns = [normalizar_nome_coluna(coluna) for coluna in df.columns]

mapa_renomeacao = {"condicao_meteorologica": "condicao_metereologica"}
df = df.rename(columns={k: v for k, v in mapa_renomeacao.items() if k in df.columns})
print(df.columns.tolist())
```

```python
# Passo 6 — Conferir se as colunas esperadas realmente vieram no arquivo
# Serve como um alarme cedo: se a PRF mudar algum nome de coluna no futuro,
# essa checagem avisa antes de qualquer etapa quebrar mais na frente.

colunas_esperadas = [
    "data_inversa", "dia_semana", "horario", "uf", "br", "municipio",
    "causa_acidente", "tipo_acidente", "classificacao_acidente",
    "fase_dia", "condicao_metereologica", "tipo_pista", "tracado_via",
    "uso_solo", "pessoas", "mortos", "feridos_leves",
    "feridos_graves", "feridos", "veiculos"
]
colunas_ausentes = [c for c in colunas_esperadas if c not in df.columns]
print("Colunas faltantes:", colunas_ausentes)
if colunas_ausentes:
    print("Atenção: ajuste nomes ou confirme o dicionário oficial da PRF usado no arquivo.")
```

```python
# Passo 7 — Primeiro retrato da base já com as colunas padronizadas

print("Dimensões:", df.shape)
print("Linhas:", df.shape[0])
print("Colunas:", df.shape[1])
print(df.head())
print(df.sample(5, random_state=42))
```

## Diagnóstico de qualidade dos dados

Antes de qualquer transformação, vale entender o estado bruto da base: quais tipos de dado vieram, quanto está faltando e se existem linhas duplicadas.

```python
# Parte 2 — Diagnóstico de qualidade dos dados
#
# Passo 8 — Tipos de dado e uso de memória

df.info(memory_usage="deep")
resumo_tipos = (
    df.dtypes.astype(str)
    .value_counts()
    .rename_axis("tipo")
    .reset_index(name="qtd_colunas")
)
print(resumo_tipos)
```

```python
# Passo 9 — Mapear valores ausentes por coluna

nulos = pd.DataFrame({
    "qtd_nulos": df.isna().sum(),
    "perc_nulos": df.isna().mean() * 100
}).sort_values("perc_nulos", ascending=False)
print(nulos[nulos["qtd_nulos"] > 0])
```

```python
# Passo 10 — Checar e remover linhas duplicadas

duplicadas_encontradas = df.duplicated().sum()
print("Duplicidades exatas:", duplicadas_encontradas)
if duplicadas_encontradas > 0:
    df = df.drop_duplicates().copy()
    print("Duplicidades removidas. Nova dimensão:", df.shape)
```

```python
# Passo 11 — Cardinalidade das colunas categóricas
# Conta quantos valores distintos cada coluna de texto tem. Colunas com
# cardinalidade muito alta (tipo município) já indicam que talvez não valha
# a pena usá-las direto num modelo sem agrupar antes.

def colunas_de_texto(base):
    nomes = []
    for coluna in base.columns:
        tipo = str(base[coluna].dtype)
        if tipo in ("object", "string", "str"):
            nomes.append(coluna)
    return nomes

colunas_categoricas = colunas_de_texto(df)
tabela_cardinalidade = (
    df[colunas_categoricas]
    .nunique(dropna=True)
    .sort_values(ascending=False)
    .reset_index()
)
tabela_cardinalidade.columns = ["variavel", "qtd_categorias"]

print(tabela_cardinalidade.head(30))
print("Cardinalidade revisada para variáveis textuais.")
```

## Transformações e variáveis derivadas

Com a base já diagnosticada, esta parte cria as variáveis de tempo (turno, faixa horária, fim de semana) e ajusta os tipos numéricos que serão usados nas análises seguintes.

```python
# Passo 12 — Converter colunas numéricas
#
# A coluna km vem com vírgula decimal (ex.: "513,8"), formato brasileiro.
# Se não trocarmos por ponto antes do pd.to_numeric, cerca de 37% das linhas
# virariam nulo só por causa do formato — não porque o dado realmente falta.

colunas_para_numerico = [
    "br", "km", "pessoas", "mortos", "feridos", "feridos_leves",
    "feridos_graves", "ilesos", "ignorados", "veiculos"
]

if "km" in df.columns:
    df["km"] = df["km"].astype(str).str.replace(",", ".", regex=False)

for coluna in colunas_para_numerico:
    if coluna in df.columns:
        df[coluna] = pd.to_numeric(df[coluna], errors="coerce")

print(df[[c for c in colunas_para_numerico if c in df.columns]].dtypes)
print("Nulos gerados em km após conversão:", df["km"].isna().sum())
```

```python
# Passo 13 — Tratar a data e derivar variáveis de tempo

df["data_inversa"] = pd.to_datetime(df["data_inversa"], errors="coerce")
df["ano"] = df["data_inversa"].dt.year
df["mes"] = df["data_inversa"].dt.month
df["trimestre"] = df["data_inversa"].dt.quarter
df["dia_semana_num"] = df["data_inversa"].dt.dayofweek
df["fim_de_semana"] = df["dia_semana_num"].isin([5, 6]).astype(int)

print(df[["data_inversa", "ano", "mes", "trimestre", "dia_semana_num", "fim_de_semana"]].head())
```

```python
# Passo 14 — Extrair a hora do campo horário e classificar o turno

horario_limpo = df["horario"].astype(str).str.strip()
df["hora"] = pd.to_datetime(horario_limpo, format="%H:%M:%S", errors="coerce").dt.hour

faltou_hora = df["hora"].isna()
if faltou_hora.any():
    df.loc[faltou_hora, "hora"] = pd.to_datetime(
        horario_limpo[faltou_hora], format="%H:%M", errors="coerce"
    ).dt.hour

def classificar_turno(hora):
    if pd.isna(hora):
        return "IGNORADO"
    if 0 <= hora <= 5:
        return "MADRUGADA"
    if 6 <= hora <= 11:
        return "MANHA"
    if 12 <= hora <= 17:
        return "TARDE"
    return "NOITE"

df["turno"] = df["hora"].apply(classificar_turno)
print(df[["horario", "hora", "turno"]].head())
```

```python
# Passo 15 — Agrupar a hora em blocos de 3 horas (faixa horária)

def criar_faixa_horaria(hora):
    if pd.isna(hora):
        return "IGNORADO"
    inicio = int(hora // 3) * 3
    fim = inicio + 2
    return f"{inicio:02d}h-{fim:02d}h"

df["faixa_horaria"] = df["hora"].apply(criar_faixa_horaria)
print(df["faixa_horaria"].value_counts(dropna=False).sort_index())
```

```python
# Passo 16 — Padronizar texto: tirar espaço nas pontas e deixar tudo maiúsculo
# Evita duplicidade de categoria só por causa de caixa alta/baixa
# (ex.: "chuva" e "Chuva" virando duas categorias diferentes).

colunas_texto = colunas_de_texto(df)
for coluna in colunas_texto:
    df[coluna] = (
        df[coluna]
        .astype("string")
        .str.strip()
        .str.upper()
    )
    df[coluna] = df[coluna].replace({"": pd.NA, "NAN": pd.NA, "NONE": pd.NA, "NULL": pd.NA})

print("Colunas textuais padronizadas:", len(colunas_texto))
```

```python
# Passo 17 — Preencher nulos das colunas categóricas mais relevantes
# Usamos "IGNORADO" em vez de descartar a linha, pra não perder registro
# só porque uma coluna secundária veio vazia.

categoricas_importantes = [
    "uf", "municipio", "causa_acidente", "tipo_acidente", "fase_dia",
    "condicao_metereologica", "tipo_pista", "tracado_via", "uso_solo",
    "classificacao_acidente", "dia_semana"
]
for coluna in categoricas_importantes:
    if coluna in df.columns:
        df[coluna] = df[coluna].fillna("IGNORADO")

print(df[categoricas_importantes].isna().sum().sort_values(ascending=False))
```

```python
# Passo 18 — Preencher nulos das colunas de contagem com zero
# Faz sentido aqui porque ausência de registro nessas colunas normalmente
# significa "não houve", não "não sabemos".

contagens_vitimas = ["mortos", "feridos", "feridos_leves", "feridos_graves", "pessoas", "veiculos"]
for coluna in contagens_vitimas:
    if coluna in df.columns:
        df[coluna] = df[coluna].fillna(0)

print(df[[c for c in contagens_vitimas if c in df.columns]].isna().sum())
```

## Construção da variável-alvo e indicadores de gravidade

Com a base já limpa, esta parte cria a variável-alvo `acidente_fatal` e alguns indicadores auxiliares de gravidade, que ajudam a enxergar o problema antes de partir para os rankings.

```python
# Passo 19 — Criar a variável-alvo acidente_fatal
# Regra: 1 quando houve pelo menos 1 morto no acidente, 0 caso contrário.

df["acidente_fatal"] = np.where(df["mortos"] >= 1, 1, 0)
validacao_alvo = (
    df["acidente_fatal"].value_counts(dropna=False)
    .rename_axis("acidente_fatal").reset_index(name="qtd")
)
validacao_alvo["perc"] = validacao_alvo["qtd"] / validacao_alvo["qtd"].sum() * 100
print(validacao_alvo)
```

```python
# Passo 20 — Validar se a regra da variável-alvo foi aplicada corretamente
# Um "cinto de segurança": se essa checagem falhar, é sinal de bug na etapa
# anterior, e o assert interrompe o notebook antes de propagar o erro.

violacoes = df.loc[
    ((df["mortos"] >= 1) & (df["acidente_fatal"] != 1)) |
    ((df["mortos"] == 0) & (df["acidente_fatal"] != 0))
]
print("Violações da regra do alvo:", len(violacoes))
assert len(violacoes) == 0, "Há erro na criação de acidente_fatal."
```

```python
# Passo 21 — Criar indicadores auxiliares de gravidade
# Esses indicadores não substituem a variável-alvo, servem só de apoio
# para a análise exploratória e o dashboard.

df["total_vitimas"] = df["mortos"] + df["feridos_leves"] + df["feridos_graves"]
df["acidente_grave"] = np.where(
    (df["mortos"] >= 1) | (df["feridos_graves"] >= 1),
    1,
    0
)
df["indice_gravidade"] = (
    df["mortos"] * 3
    + df["feridos_graves"] * 2
    + df["feridos_leves"]
)
print(df[["mortos", "feridos_leves", "feridos_graves", "total_vitimas", "indice_gravidade"]].head())
```

```python
# Passo 22 — Padronizar o número da BR e montar uma chave de localidade
# A chave (UF + município + BR) ajuda a agrupar/comparar trechos específicos
# mais na frente, sem precisar repetir três filtros toda vez.

def formatar_br(valor):
    if pd.isna(valor) or valor == 0:
        return "BR-IGNORADA"
    return f"BR-{int(valor):03d}"

df["br_formatada"] = df["br"].apply(formatar_br)
df["chave_localidade"] = (
    df["uf"].astype(str) + "_" +
    df["municipio"].astype(str) + "_" +
    df["br_formatada"].astype(str)
)
print(df[["uf", "municipio", "br", "br_formatada", "chave_localidade"]].head())
```

```python
# Passo 23 — Conferência rápida dos números depois de todas as transformações

checagens = {
    "linhas": len(df),
    "colunas": df.shape[1],
    "acidentes_fatais": int(df["acidente_fatal"].sum()),
    "taxa_fatalidade": float(df["acidente_fatal"].mean()),
    "total_mortos": int(df["mortos"].sum()),
    "total_feridos": int(df["feridos"].sum()) if "feridos" in df.columns else None,
}
checagens
```

```python
# Passo 24 — Ranking rápido das categorias mais frequentes

def ranking_categoria(base, coluna, n=10):
    return (
        base[coluna]
        .value_counts(dropna=False)
        .head(n)
        .rename_axis(coluna)
        .reset_index(name="qtd")
    )

print(ranking_categoria(df, "causa_acidente", 10))
print(ranking_categoria(df, "tipo_acidente", 10))
```

```python
# Passo 25 — Taxa de fatalidade por categoria
# Corte mínimo de registros pra não deixar categoria rara distorcer o
# percentual (mesmo critério usado nas consultas SQL do Módulo 3).

def taxa_fatal_por_categoria(base, coluna, min_registros=30):
    tab = base.groupby(coluna).agg(
        qtd_acidentes=("acidente_fatal", "size"),
        qtd_fatais=("acidente_fatal", "sum"),
        taxa_fatal=("acidente_fatal", "mean")
    ).reset_index()
    tab = tab[tab["qtd_acidentes"] >= min_registros]
    return tab.sort_values("taxa_fatal", ascending=False)

print(taxa_fatal_por_categoria(df, "tipo_acidente", min_registros=30).head(10))
```

```python
# Passo 26 — Gráfico de barras da distribuição do alvo
# Só uma conferência visual do desbalanceamento entre fatais e não fatais.

ax = df["acidente_fatal"].value_counts().sort_index().plot(kind="bar")
ax.set_title("Distribuição da variável-alvo acidente_fatal")
ax.set_xlabel("acidente_fatal")
ax.set_ylabel("Quantidade de acidentes")
plt.show()
```

## Construção das bases finais e documentação

Última etapa: gerar a base analítica completa, selecionar as variáveis modeláveis (sem vazamento de dado), exportar os arquivos e registrar as decisões tomadas.

```python
# Passo 27 — Gerar a base analítica completa
# Essa base guarda tudo, inclusive as colunas de desfecho — é usada para
# EDA e Power BI, não para o modelo.

base_analitica = df.copy()
print("Base analítica:", base_analitica.shape)
print("Colunas:", base_analitica.columns.tolist())
```

```python
# Passo 28 — Selecionar as variáveis que vão para a base modelável
# Aqui já ficam de fora as colunas de desfecho (mortos, feridos etc.);
# a conferência final de vazamento acontece no próximo passo.

variaveis_modelaveis = [
    "uf", "br_formatada", "municipio", "mes", "trimestre",
    "dia_semana", "dia_semana_num", "fim_de_semana",
    "hora", "faixa_horaria", "turno", "fase_dia",
    "causa_acidente", "tipo_acidente", "condicao_metereologica",
    "tipo_pista", "tracado_via", "uso_solo",
    "acidente_fatal"
]
variaveis_modelaveis = [c for c in variaveis_modelaveis if c in df.columns]
base_modelavel = df[variaveis_modelaveis].copy()
print("Base modelável:", base_modelavel.shape)
```

```python
# Passo 29 — Confirmar que nenhuma variável de desfecho entrou na base modelável
# Se qualquer uma dessas colunas aparecer aqui, é sinal de vazamento de dado,
# e o processo para com erro em vez de seguir silenciosamente.

variaveis_proibidas = [
    "mortos", "feridos", "feridos_leves", "feridos_graves",
    "total_vitimas", "indice_gravidade", "acidente_grave",
    "classificacao_acidente"
]

def verificar_data_leakage(base, proibidas):
    presentes = [c for c in proibidas if c in base.columns]
    if presentes:
        raise ValueError(f"Data leakage detectado: {presentes}")
    return "OK — nenhuma variável proibida encontrada."

verificar_data_leakage(base_modelavel, variaveis_proibidas)
```

## Fechamento: tratamento final, exportação e documentação

As etapas acima já cobriram limpeza, criação de variáveis e separação das bases. Falta só o tratamento final de nulos na base modelável, a exportação dos arquivos e o registro da documentação de apoio (dicionário, decisões e README).

```python
# Passo 30 — Tratamento final de nulos na base modelável
# A variável-alvo fica de fora de propósito: ela não deveria ter nulo a
# essa altura, e preenchê-la esconderia um problema real se houvesse.

for coluna in base_modelavel.columns:
    if coluna == "acidente_fatal":
        continue
    if base_modelavel[coluna].dtype == "object" or str(base_modelavel[coluna].dtype) == "string":
        base_modelavel[coluna] = base_modelavel[coluna].fillna("IGNORADO")
    else:
        base_modelavel[coluna] = base_modelavel[coluna].fillna(-1)

print(base_modelavel.isna().sum().sort_values(ascending=False).head())
```

```python
# Passo 31 — Exportar as duas bases tratadas em CSV

base_analitica.to_csv(ARQUIVO_BASE_ANALITICA, index=False, sep=SEPARADOR, encoding=ENCODING_SAIDA)
base_modelavel.to_csv(ARQUIVO_BASE_MODELAVEL, index=False, sep=SEPARADOR, encoding=ENCODING_SAIDA)

print("Arquivos exportados:")
print("-", ARQUIVO_BASE_ANALITICA)

print("-", ARQUIVO_BASE_MODELAVEL)
print("Estrutura final pronta para uso em análise e modelagem.")
```

```python
# Passo 32 — Reabrir os arquivos exportados como conferência
# Garante que o CSV gravado bate exatamente com o que estava em memória,
# pegando qualquer problema de separador/encoding antes de seguir.

valid_analitica = pd.read_csv(ARQUIVO_BASE_ANALITICA, sep=SEPARADOR, encoding=ENCODING_SAIDA)
valid_modelavel = pd.read_csv(ARQUIVO_BASE_MODELAVEL, sep=SEPARADOR, encoding=ENCODING_SAIDA)

print("Analítica reaberta:", valid_analitica.shape)
print("Modelável reaberta:", valid_modelavel.shape)
assert len(valid_analitica) == len(base_analitica)
assert len(valid_modelavel) == len(base_modelavel)
```
